```python
import pandas as pd
import numpy as np
```

```python
# Series
s = pd.Series([21,19,20])

# Access using index (Position)
print(s[0])

# Giving custom index (named Index)
s= pd.Series([21,19,20],index = ["Aryan","Nitin","Sneha"])

# Access using named index
print(s["Sneha"])

# Access using iloc for position index
print(s.iloc[0]) # Single Value
print(s.iloc[[0,1]]) # Multiple Value

# Access using loc for the custom index
print(s.loc["Sneha"]) # Single Label
print(s.loc[["Aryan","Sneha"]]) # Multiple Label

# To print all indexes
print(s.index)
```

```python
# Properties of Series

# Homogenous - it keeps the data of same data type
s = pd.Series([24,25,26,27,"Aryan"])
print(s)

# Vectorized Operation - can use multiple operations on Series without using loops
s1 = pd.Series([5,6,7,8])
s2 = pd.Series([10,20,30,40])
sum_series = s1 + s2
print(sum_series)

# Handles missing values using NaN - it handles missing values using NaN

# mutable values - can change the values of Series
s = pd.Series([24,25,26])
s[0] = 99
print(s)

# Immutable size - cannot change the size of the Series

s = pd.Series([24,25,16,12])
changed_series = s.drop(0)
print(s)
print(changed_series)
```

```python
# DataFrame Creation

# Using Dictionary
info = {
    "Name":["Aryan","Sneha"],
    "age":[21,20],
}
df = pd.DataFrame(info)
print(df)
# To access row 
print(df.index)
# To access columns
print(df.columns)

# Using List of List
df_list = [["Aryan","Nitin","Sneha"],[21,19,20],[9.6,9.2,9.9]]
df = pd.DataFrame(df_list,columns=["Name","age","cgpa"])
print(df)

# Using np.arrays
np_arr = np.array([["Aryan","Nitin","Sneha"],[21,19,20],[9.6,9.2,9.9]])
df = pd.DataFrame(np_arr,columns = ["Name","age","cgpa"])
print(df)
```

```python
df = pd.read_csv("globalAirQuality.csv")
df
```

```python
# Selecting and Indexing Data

# Columns
df[["country","aqi"]]

# Rows
#Using loc (for the labelled index)
df.loc[22]
df.loc[0:2] # Ending index is inclusive

#Using iloc (for position index)
df.loc[1]
df.iloc[0:2] # Ending index is exclusive

# Selecting rows and columns together
#Using loc (for the labelled index)
df.loc[0:2,["country","city","aqi"]] # Ending index is inclusive

#Using iloc (for position index)
df.iloc[0:2,2:4] # Ending index is exclusive


# Single scalar value

# Using at (for the labelled index)
df.at[0,"city"]

#Using iat (for the position index)
df.iat[0,2]
```

```python
# Filtering (Searching data based on the conditions)

df[df["aqi"] > 100]
df[(df["aqi"] > 100)  & (df["city"] == "Delhi")]

filtered_data = df[(df["aqi"] > 100)  & (df["city"] == "Delhi")][["city","country"]]
filtered_data.iloc[0]
```

```python
# Query Method (always return copy of data)
df.query("aqi > 100  & temperature > 30")

# to use attribute use @
city = "Delhi"
df.query("aqi > 100 & city == @city ")
```

```python
# Data Handling 
df = pd.read_csv("raw_data.csv")

# Handling Missing Values

# isnull() and isna() :- To check whether the values are null or not ( na = not available)
df.isnull()
df.isna()

# isnull.sum() :- To check the number of missing values in each column
df.isnull().sum()

# dropna() :- To remove missing rows
df.dropna()

# fillna(value) :- To fill the missing value with a value

cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()

cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
cleaned_data

# forward fill
df.ffill()

# backward fill
df.bfill()
```

```python
# Handling duplicate values

# To find duplicates
df["name"].duplicated()

df2 = df.copy()

# To remove duplicates
df.drop_duplicates()
df.drop_duplicates(inplace = True) # To make changes in the original data frame
df
```

```python
# Data tupes 

df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64") # To change the data types of a perticular column
df2.dtypes # To check the data types of each columns

# to_datatime(date ) 
s = pd.Series([pd.to_datetime("2026-06-30")])
type(s.dtypes)
```

```python
# Handling Strings

# To make lower case
df2["country"] = df2["country"].str.lower()

# To make Upper case
df2["country"] = df2["country"].str.upper()

# To split
df2["name"].str.split(" ")

# To check whether value exists or not . also it allow case insensitive
df2["name"].str.contains("alex",case = False)
```
