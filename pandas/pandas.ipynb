```python
import pandas as pd
import numpy as np
```

```python
# Series
s = pd.Series([21,19,20])

# Access using index (Position)
print(s[0])

# Giving custom index (named Index)
s= pd.Series([21,19,20],index = ["Aryan","Nitin","Sneha"])

# Access using named index
print(s["Sneha"])

# Access using iloc for position index
print(s.iloc[0]) # Single Value
print(s.iloc[[0,1]]) # Multiple Value

# Access using loc for the custom index
print(s.loc["Sneha"]) # Single Label
print(s.loc[["Aryan","Sneha"]]) # Multiple Label

# To print all indexes
print(s.index)
```

```python
# Properties of Series

# Homogenous - it keeps the data of same data type
s = pd.Series([24,25,26,27,"Aryan"])
print(s)

# Vectorized Operation - can use multiple operations on Series without using loops
s1 = pd.Series([5,6,7,8])
s2 = pd.Series([10,20,30,40])
sum_series = s1 + s2
print(sum_series)

# Handles missing values using NaN - it handles missing values using NaN

# mutable values - can change the values of Series
s = pd.Series([24,25,26])
s[0] = 99
print(s)

# Immutable size - cannot change the size of the Series

s = pd.Series([24,25,16,12])
changed_series = s.drop(0)
print(s)
print(changed_series)
```

```python
# DataFrame Creation

# Using Dictionary
info = {
    "Name":["Aryan","Sneha"],
    "age":[21,20],
}
df = pd.DataFrame(info)
print(df)
# To access row 
print(df.index)
# To access columns
print(df.columns)

# Using List of List
df_list = [["Aryan","Nitin","Sneha"],[21,19,20],[9.6,9.2,9.9]]
df = pd.DataFrame(df_list,columns=["Name","age","cgpa"])
print(df)

# Using np.arrays
np_arr = np.array([["Aryan","Nitin","Sneha"],[21,19,20],[9.6,9.2,9.9]])
df = pd.DataFrame(np_arr,columns = ["Name","age","cgpa"])
print(df)
```

```python
# To read data from different format of files using pandas

# For csv(comma-seperated values) format file
df = pd.read_csv("employee_data.csv")
print(df)

# For json(JavaScript Object Notation) format file
df = pd.read_json("employee_data.json")
print(df)
```

```python
# Operations on DataFromes 

# Head : To get the first five entries of the top of the data as default or can specify number
df.head()
df.head(6)

# Tail : To get the Last five entries of the last of the data as default or can specify number
df.tail()
df.tail(10)

# Sample : To get the random number of entries from the data 
df.sample()
df.sample(4)

# Info : To get the information about the DataFrame
df.info()

# Shape : To get the dimension of the DataFrame (rows,columns)
df.shape

# Describe : To do the numerical data from the given DataFrame
df.describe()

# columns : To get all columns
df.columns

# nunique : To get n unique values from the DataFrame
df.nunique()
```
