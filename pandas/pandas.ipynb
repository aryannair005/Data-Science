```python
import pandas as pd
import numpy as np
```

```python
# Series
s = pd.Series([21,19,20])

# Access using index (Position)
print(s[0])

# Giving custom index (named Index)
s= pd.Series([21,19,20],index = ["Aryan","Nitin","Sneha"])

# Access using named index
print(s["Sneha"])

# Access using iloc for position index
print(s.iloc[0]) # Single Value
print(s.iloc[[0,1]]) # Multiple Value

# Access using loc for the custom index
print(s.loc["Sneha"]) # Single Label
print(s.loc[["Aryan","Sneha"]]) # Multiple Label

# To print all indexes
print(s.index)
```

```python
# Properties of Series

# Homogenous - it keeps the data of same data type
s = pd.Series([24,25,26,27,"Aryan"])
print(s)

# Vectorized Operation - can use multiple operations on Series without using loops
s1 = pd.Series([5,6,7,8])
s2 = pd.Series([10,20,30,40])
sum_series = s1 + s2
print(sum_series)

# Handles missing values using NaN - it handles missing values using NaN

# mutable values - can change the values of Series
s = pd.Series([24,25,26])
s[0] = 99
print(s)

# Immutable size - cannot change the size of the Series

s = pd.Series([24,25,16,12])
changed_series = s.drop(0)
print(s)
print(changed_series)
```

```python
'''Feature engineering :Feature engineering is the process of creating, transforming, and selecting features
from raw or existing data to make them more useful for machine learning algorithms.'''

df = pd.read_csv("raw_data.csv")
df = df.fillna(0)
df2 = df.copy()
df2

# apply(fn) : To apply a custom function to each value of Series or DataFrame  (We write function inside the apply function)
# Creating new tax column if income > 50000 then tax = 20% else tax = 10% 

df2["tax"] = df2["income"].apply(lambda x : x*0.2 if x > 50000 else x*0.1)
df2

# map() : When we want to map a value to the existing value 
gender_map = {"Male":"M","Female":"F","Unknown":"U"}
df2["gender"] = df2["gender"].map(gender_map)
df2

# assign() : Can create new column by assigning them a value
df2 = df2.assign(new_income = df2["income"] + df2["tax"])
df2

# replace() : To replace old value with a new value
df2["country"]  = df2["country"].replace("USA","US")
df2

# rename(column = {old : new}) : To rename a existing column or row(index) name

# To rename all column names
df2.columns = ["Id","Name","Age","Country","Gender","Income","Tax","New_Income"]

# To change a perticular column name
df2.rename(columns = {"Income":"Salary"})

# To change a perticular row
df2.rename(index = {0:"First"})
df2

# sort_values : To sort a column based on the value it contain
df2.sort_values("Income")

# Decending order
df2.sort_values("Income",ascending = False)

# To sort index
df2.sort_index()

# Reset index
df2.reset_index() # This will also preserve the old index
df2.reset_index(drop = True) # This will not preserve the old index

# Ranking : Ranking the column based on the value . on ties (having same value it will calculate the average)

df2["Ranking"] = df2["Income"].rank() # Lowest value get the rank 1
df2["Ranking"] = df2["Income"].rank(ascending = False) # Highest value get the rank 1
df2["Ranking"] = df2["Income"].rank(ascending = False,method = "dense") # To get a whole value of rank
# To resolve the tie problem
df2["Ranking"] = df2["Income"].rank(ascending = False,method = "min") # To assign the lowest value on tie
df2["Ranking"] = df2["Income"].rank(ascending = False,method = "max") # To assign the Highest value on tie

df2

# To reorder columns
df2[["Id","Name","Age","Gender","Country","Income","Tax","New_Income"]]

# Practice question to shift the id column to the end using list comprehension

new_order = [col for col in df2.columns  if col != "Id"  ] + ["Id"]
df2[new_order ]
```

```python
### Melt & Pivot

df = pd.DataFrame({
    "country" : ["USA","USA","India","India"],
    "year" : [2020,2021,2020,2021],
    "sales" : [100,120,90,110],
    "profit" : [20,25,18,22],
})

# Melt
''' Melt table : Wide to long Format . we pass 4 arguments.
1 - id_vars = The columns you want to preserve as it is.
2 - value_vars = The columns you want to melt.
3 - var_name = The melted column variable name
4 - value_name = The melter column value name ''' 

melted_data = df.melt(
    id_vars = ["country","year"],
    value_vars = ["sales","profit"],
    var_name = "metrics",
    value_name = "value"
)
melted_data

# Pivot
''' Pivot table : Long to Wide Format . we pass 3 arguments.
1 - index : unique list of columns
2 - columns : column name you want to pivot
3 - values : value of the pivot column '''

original_value = melted_data.pivot(
    index = ["country","year"],
    columns = "metrics",
    values = "value"
)
original_value
```

```python
# Merging and Joining data

df_customers = pd.DataFrame({
    "customer_id":[1,2,3,4],
    "name":["Adam","Bob","Charlie","Dave"]
})
df_orders = pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,300,180]
})

# Inner Join 
pd.merge(df_customers,df_orders,on = "customer_id") 

# Left Join
pd.merge(df_customers,df_orders,on = "customer_id",how ="left")

# Right Join
pd.merge(df_customers,df_orders,on = "customer_id",how ="right")

# Outer Join
pd.merge(df_customers,df_orders,on = "customer_id",how ="outer")
```

```python
# Data Concatenation
df_customers1 = pd.DataFrame({
    "id":[1,2,3,4],
    "name":["Adam","Bob","Charlie","Dave"]
})
df_customers2 = pd.DataFrame({
    "id":[5,6,7,8],
    "name":["Aryan","Sneha","Nitin","Sanjay"]
})
# Row Concatination
pd.concat([df_customers1,df_customers2])
pd.concat([df_customers1,df_customers2],ignore_index = True) # To not to preserve index

# Column Concatination
pd.concat([df_customers1,df_customers2],axis = 1)
```
