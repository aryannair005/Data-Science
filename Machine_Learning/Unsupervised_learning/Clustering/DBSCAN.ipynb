```python
from sklearn.datasets import load_iris,make_moons
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
```

```python
iris_data = load_iris()
```

```python
X = iris_data.data
y = iris_data.target
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
sns.scatterplot(x = X_scaled[:,0] , y = X_scaled[:,2] , c = y)
```

```python
# DBSCAN : For Linear Data
db_scan = DBSCAN(
    eps = 0.5,
    min_samples = 5
)
labels = db_scan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x = X_scaled[:,0] , y = X_scaled[:,2] , c = labels)
```

```python
# For non linear data 
X , y = make_moons(
    n_samples = 300,
    noise = 0.05
)
X_scaled = scaler.fit_transform(X)
```

```python
sns.scatterplot(x = X_scaled[:,0] , y = X_scaled[:,1] , c = y)
```

```python
db_scan = DBSCAN(
    eps = 0.5,
    min_samples = 5
)
labels = db_scan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x = X_scaled[:,0] , y = X_scaled[:,1] , c = labels)
```
