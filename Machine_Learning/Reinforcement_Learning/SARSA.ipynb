```python
import gymnasium as gym
import random
import numpy as np
```

```python
# Sample Environment(env)
env = gym.make("CliffWalking-v1")
```

```python
# To see number of states
no_of_states = env.observation_space.n
print(no_of_states)

# To see number of actions
no_of_actions = env.action_space.n
print(no_of_actions)

# Starting state 
starting_state , _ = env.reset() # row(3) * columns(12) + column(0) => 36
print(starting_state)
```

# SARSA

### Hyperparameters

```python
gamma = 0.99
alpha = 0.5
epsilon = 0.1
episodes = 500
```

### Q-Table

```python
Q = np.zeros((48,4))  #(number of states(12*4) , number of actions)
```

### Policy (Strategy to choose action)

```python
# Policy - epsilon-greedy : state -> action

def epsilon_greedy(state):
    if random.random() < epsilon:
        return env.action_space.sample() # Random Action => Explore
    else:
        return np.argmax(Q[state])
```

### Training SARSA

```python
for episode in range(episodes):
    if (episode % 50 == 0):
        env = gym.make("CliffWalking-v1", render_mode = "human")
    else:
        env = gym.make("CliffWalking-v1") # Creating a new environment

    done = False # To check the termination or truncate condition
    
    state , _ = env.reset() # Starting state 
    action = epsilon_greedy(state) # Choosing action based on the current state

    total_reward = 0 
    episode_len = 0

    while not done:
        next_state , reward , terminated , truncated , _ = env.step(action) # Taking action
        done = terminated or truncated

        if done:
            # Termination Condition
            # Sarsa Update 
            Q[state, action] += alpha * (reward - Q[state, action])
        else:
            next_action = epsilon_greedy(next_state)
            # Sarsa Update 
            Q[state, action] += alpha * (reward + gamma * Q[next_state, next_action] - Q[state, action])

            state = next_state
            action = next_action

        total_reward += reward 
        episode_len += 1

    print(f"Episode = {episode+1}/500 : total reward = {total_reward} & ep length = {episode_len}")
    env.close()
```

### What did agent learned ?

```python
env = gym.make("CliffWalking-v1", render_mode = "human")

state , _ = env.reset()
done = False
total_reward = 0
episode_len = 0

while not done:
    action = np.argmax(Q[state])
    next_state , reward , terminated, truncated , _ = env.step(action)

    done = terminated or truncated
    state = next_state

    total_reward += reward 
    episode_len += 1

print(f"total reward : {total_reward} & episode len : {episode_len} ")
env.close()
```
