```python
import gymnasium as gym 
import numpy as np
import random
```

### Hyperparameters

```python
gamma = 0.99
epsilon = 0.1 
alpha = 0.5
episodes = 500
```

### Q-values 

```python
Q = np.zeros((48,4))  #(number of states(12*4) , number of actions)
```

### Policy

```python
def epsilon_greedy(state):
    if random.random() < epsilon:
        return env.action_space.sample()

    else:
        return np.argmax(Q[state])     
```

# Q-Learning

```python
for episode in range(episodes):
    render = (episode % 100 == 0)

    if render:
        env = gym.make("CliffWalking-v1",render_mode = "human")
    else:
        env = gym.make("CliffWalking-v1")


    done = False
    state , _ = env.reset()

    episode_len = 0
    tot_reward = 0

    while not done:
        action = epsilon_greedy(state)

        next_state , reward , terminated , truncated , _ = env.step(action)
        done = terminated or truncated

        # Q-learning updation
        Q[state , action ]  += alpha * (reward + gamma * np.max(Q[next_state]) - Q[state,action])

        state = next_state
        episode_len += 1 
        tot_reward += reward

    print(f"for episode : {episode+1}  reward : {reward}  & episode len {episode_len}")
    env.close()
```

### What did agent learned ?

```python
env = gym.make("CliffWalking-v1",render_mode = "human")

state , _ = env.reset()
tot_reward = 0
episode_len = 0
done = False

while not done:
    action = np.argmax(Q[state])
    state , reward , terminated , truncated , _ = env.step(action)
    done = terminated or truncated


    episode_len += 1
    tot_reward += reward

print(f"total reward : {tot_reward} & episode length : {episode_len}")
env.close()
```
