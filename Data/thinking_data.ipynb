```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("store_data.json")
print(data)
print(type(data))
```

```python
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[]
    unique_user=set()
    
    for user in data:
        # Clear rating (data consistency)
        raw_rating=str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating=text_to_num[raw_rating]

        user["rating"]=float(raw_rating)

        # Missing Values 
        raw_age=user.get("age")
        if(raw_age == None):
            user["age"]=None

        # Duplicate Data (Deduplication)
        raw_name=user["name"].strip().lower()
        if raw_name in unique_user:
            continue
        else:
            unique_user.add(raw_name)
            user["name"] =raw_name
            cleaned_data.append(user)


    return cleaned_data

data= clean_data(data)
```

```python
# Giving meaningful insights
def get_insights(data):
    total_rating=0.0
    bad_rating=0
    for user in data:
        if(user["rating"] != None):
            if(user["rating"]<3):
                bad_rating+=1
            total_rating+=user["rating"]

    print(f"average rating : {int(total_rating/len(data))}") 
    print(f"bad rating : {bad_rating}")

get_insights(data)
```

```python
# Recommendation
def getRecommendation(data):
    recommendation=[]
    for user in data:
        curr_recommendation={}
        curr_recommendation["name"]=user["name"]
        if(user["rating"] >=4):
            curr_recommendation["brand"]="apple"
        else:
            curr_recommendation["brand"]="samsung"
        recommendation.append(curr_recommendation)

    return recommendation

getRecommendation(data)
```
