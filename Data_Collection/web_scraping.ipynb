```python
import requests
```

```python
# Sending request using requests module
URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
if res.status_code == 200:
    print("Request sent successfully")
else:
    print("Error")
```

```python
# Storing data after sending request
with open("scraped_data/data1.html","w",encoding="utf-8") as f:
    f.write(res.text)
```

```python
# BeautifulSoup (after storing it we create a soup which create a parse tree to extract usefull information)
from bs4 import BeautifulSoup
```

```python
# Open the file to make soup of data
with open("scraped_data/data1.html","r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content,"lxml")
```

```python
# Extracting meaning insights
all_h3 = soup.find_all("h3")
all_info = []
for h3 in all_h3:
    name = h3.get_text(strip = True)
    population = h3.find_next("div").select("span.country-population")[0].get_text()
    all_info.append([name , population])

all_info
```

```python
# Storing the extracted usefull data
import pandas as pd
```

```python
df = pd.DataFrame(all_info,columns=["country-name","population"])
df.to_csv("cleaned_data/data.csv",index = False)
```
