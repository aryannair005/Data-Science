```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("store_data.json")
print(data)
```

```python
#Clean and Structure data
def clean_data(data):
    cleaned_data = []
    unique_users = set()
    
    text_num = {"one":1,"two":2,"three":3,"four":4,"five":5}
    for user in data:
        count+=1
        
        # Clean rating
        raw_rating = str(user["rating"]).strip().lower()
        
        if(raw_rating in text_num):
            raw_rating = text_num[raw_rating]
        user["rating"] = raw_rating

        # Missing values
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # Deduplication
        if(user["name"].strip().lower() in unique_users):
            continue

        unique_users.add(user["name"].strip().lower())
        cleaned_data.append(user)

    return cleaned_data
        
    
```

```python
# Get meaningfull insights 
def averageRating(data):
    average = 0
    count = 0
    for user in data:
        count+=1
        average += float(user["rating"])

    return average/count

def poorRating(data):
    poor_rating_count = 0
    total_user_count = 0
    for user in data:
        total_user_count += 1
        user_rating = float(user["rating"])
        if(user_rating < 3):
            poor_rating_count += 1

    return (poor_rating_count/total_user_count) * 100
    
```

```python
poorRating(data)
```

```python
# Recommendation 
def get_recommendations(data):
    recommendations = []
    for user in data:
        curr_recommendation = {}
        curr_recommendation["name"] = user["name"]

        if(float(user["rating"]) < 4):
            curr_recommendation["brand"] = "samsung"
        else:
            curr_recommendation["brand"] = "apple"

        recommendations.append(curr_recommendation)

    return recommendations
```

```python
get_recommendations(data)
```
