```python
# Real Data
oscar_movies = [
    "Bhaag Milkha Bhaag",
    "3 idiots",
    "12th fail",
    "Zindagi Na Milegi Dobara",
    "Shershaah"
]
years = [2012,2013,2014,2015,2016]
oscar_revenue = [100,200,300,400,500]

plt.plot(years,oscar_revenue)

# To give title to the plot
plt.title("Oscar movies revenue in each year")

# X-axis label
plt.xlabel("Years")

# Y-axis label
plt.ylabel("revenue (in crores)")
```

```python
# Multiple data plots
non_oscar_movies = [
    "Bajrangi Bhaijaan",
    "Pritam And Pedro",
    "Stree",
    "Sector 36",
    "Housefull"
]

non_oscar_revenue = [800,600,50,70,100]

plt.plot(years,oscar_revenue , label = "oscar movies")
plt.plot(years,non_oscar_revenue , label = "non-oscar movies")

plt.title("Oscar vs non-oscar Movies")
plt.xlabel("years")
plt.ylabel("revenue (in crores)")
# To show the plot label
plt.legend()
# Can specify position of the lable box
plt.legend(loc = "upper left")
```

```python
# Format strings : A format string consists of a part for color, marker and line
plt.plot(years,oscar_revenue ,"o:r",label = "oscar movies")
plt.plot(years,non_oscar_revenue ,",--b", label = "non-oscar movies")

plt.title("Oscar vs non-oscar Movies")
plt.xlabel("years")
plt.ylabel("revenue (in crores)")
plt.legend()

# can give seprate values

plt.plot(years,oscar_revenue ,color = "green" ,marker = "s",linestyle="-",linewidth = "5",label = "oscar movies")
```

```python
# Bar Graph (This creates a vertical bar graph where x-axis = categories y-axis = values)
years = [2012,2013,2014,2015,2016]
oscar_revenue = [100,200,300,400,500]

plt.bar(years,oscar_revenue)
plt.title("Oscar movies revenue each year")
plt.xlabel("years")
plt.ylabel("revenue (in crores)")

# To give bar labels (text above the bar)
# plt.text( x-axis alignment, y-axis alignment, text to display, alignment)

# For individual cell
plt.text(years[0],oscar_revenue[0] + 10,oscar_revenue[0],ha = "center")

# For all cells
for i in range(len(years)):
    plt.text(years[i],oscar_revenue[i] + 10,oscar_revenue[i],ha = "center")


# To increase x axis limit
plt.xlim()

# To increase y axis limit
plt.ylim(0,max(oscar_revenue)+100)
```

```python
# Multiple bar garphs

import numpy as np

years = [2012,2013,2014,2015,2016]
oscar_revenue = [100,200,300,400,500]
non_oscar_revenue = [800,600,50,70,100]

x = np.arange(len(years))
width = 0.35

plt.bar(x-width/2  , oscar_revenue , width , label = "oscar movies")
plt.bar(x+width/2  , non_oscar_revenue , width , label = "non-oscar movies")
plt.title("Oscar movies revenue each year")
plt.xlabel("years")
plt.ylabel("revenue (in crores)")
plt.legend()

# To replace index value with the required values 
plt.xticks(x,years)
```

```python
# Horizontal bar chart (first pass y axis - categories and then pass x axis - values)
oscar_movies = [
    "Bhaag Milkha Bhaag",
    "3 idiots",
    "12th fail",
    "Zindagi Na Milegi Dobara",
    "Shershaah"
]
oscar_revenue = [100,200,300,400,500]
plt.barh(oscar_movies,oscar_revenue)
plt.title("Oscar movies revenue")
plt.xlabel("oscar movies")
plt.ylabel("revenue (in crores)")
```

```python
# Scatter Plots

people = [
    "Person A", "Person B", "Person C", "Person D", "Person E",
    "Person F", "Person G", "Person H", "Person I", "Person J"
]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.scatter(age,bp)
plt.title("Age vs Bp")
plt.xlabel("Age")
plt.ylabel("Bp")
plt.grid()

# Properties 

# Colors , size , marker , Transparency ,  
color = ["green" if x < 120 else "red" for x in bp]
plt.scatter(age,bp,color = color, s = bp , marker = "s",alpha = 0.8) 

# color map with range
plt.scatter(age,bp,s=bp,cmap= "OrRd",c=bp)


# To know the scale of color bar
plt.colorbar(label = "BP")
```

```python
# Add Annotations
people = [
    "Person A", "Person B", "Person C", "Person D", "Person E",
    "Person F", "Person G", "Person H", "Person I", "Person J"
]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.scatter(age,bp)
plt.title("Age vs Bp")
plt.xlabel("Age")
plt.ylabel("Bp")
plt.grid()

for i in range(len(people)):
    # annotate(what u want as text , on which (x,y) point , allignment of text (x,y))
    plt.annotate(people[i],xy = (age[i],bp[i]),xytext = (age[i],bp[i]+1) )

plt.xlim(min(age),max(age)+10)
plt.ylim(100,max(bp)+10)
```

```python
# Pie charts and customization of pie charts

Software_names = ["Windows","IOS","Linux"]
users = [1000,200,50]
colors = ["lightblue","grey","yellow"]
explode_values = [0.1,0,0]

plt.pie(users , labels = Software_names,
        autopct ="%1.1f%%", # Give percentage of each part
        colors = colors,    # To give custom colors to the pie chart
        wedgeprops = {      # To give properties to a specific edge
            "edgecolor" : "black",
            "linewidth" : 2,
            "linestyle" : "--",
            "alpha" : 0.9,
            # "fill" : False  # To remove colors
        },
        explode = explode_values, # To specifically standout any cell
        startangle = 90 , # To rotate the pie chart at a specific angle
        shadow = True # To add shadow (3D) effect
       )
```
