```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# First set theme
sns.set_theme()

# Seaborn provide many datasets
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
```

```python
# For scatter plot
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "smoker",
    style = "smoker",
    size = "size"
)
```

```python
# Bar plots 
sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# Box plot 
sns.boxplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# Distribution plot
sns.histplot(
    data = tips,
    x = "total_bill",
    bins = 50
)
```

```python
flights = sns.load_dataset("flights")
```

```python
flights_pivot_data = flights.pivot(index = "month",columns = "year",values = "passengers")
```

```python
# Heat maps 
sns.heatmap(
    flights_pivot_data,
    cmap = "coolwarm" ,# Color
    annot = True,
    fmt = "d"
)
plt.title("Passengers Heatmaxp")
plt.tight_layout()
```

```python
# Best Practice for data visualization
fig, ax = plt.subplots()
sns.lineplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex",
    ax = ax
)

fig.tight_layout()

```
